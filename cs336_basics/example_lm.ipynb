```python
d_k = 6
theta = torch.pi/10

# Compute frequencies (correct formula)
k = torch.arange(0, d_k // 2)
freqs = 1.0 / (theta ** (2 * k / d_k))  # (d_k/2,)

positions = torch.arange(4).unsqueeze(1)  # (L, 1)

# Outer product → angles
angles = positions * freqs  # (L, d_k/2)
```
